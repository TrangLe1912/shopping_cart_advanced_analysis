```python
import pandas as pd
import os
import sys
import time
# Thêm đường dẫn để nhận diện thư mục src
sys.path.append(os.path.abspath('..'))
from src.apriori_library import FPGrowthMiner, DataVisualizer

# Cấu hình đường dẫn
BASKET_BOOL_PATH = "../data/processed/basket_bool.csv"
OUTPUT_DIR = "../results/fp_growth/"
os.makedirs(OUTPUT_DIR, exist_ok=True)

# Thông số (Bạn có thể sửa số này để nộp bài)
MIN_SUPPORT = 0.01 
MIN_CONFIDENCE = 0.5
```

```python
# Đọc dữ liệu giỏ hàng
basket_bool = pd.read_csv(BASKET_BOOL_PATH, index_col=0).astype(bool)

print("Đang chạy FP-Growth...")
start_time = time.time()

# Gọi class FP-Growth
miner_fp = FPGrowthMiner(basket_bool=basket_bool)
frequent_itemsets = miner_fp.run(min_support=MIN_SUPPORT)
rules_fp = miner_fp.create_association_rules(min_confidence=MIN_CONFIDENCE)

end_time = time.time()
print(f"Thời gian chạy: {end_time - start_time:.4f} giây")
print(f"Tìm thấy: {len(rules_fp)} luật kết hợp.")
rules_fp.head(10)
```
